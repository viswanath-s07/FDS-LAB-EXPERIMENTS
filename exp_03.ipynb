```python
import pandas as pd
import numpy as np

data = {
    "CustomerID": [101, 102, 103, 104, 105, 106, 107, 108, 109, 109, 110],
    "Age_Group": ["18-25", "26-35", "36-45", "18-25", "46+", "36-45", "26-35", "18-25", "36-45", "36-45", "26-35"],
    "Rating(1-5)": [5, 4, 3, -2, 4, 5, 6, 2, 3, 4, 5],
    "Restaurant": ["A2B", "Sangeetha", "Saravana Bhavan", "A2B", "Junior Kuppanna", "A2B", "Saravana Bhavan", "Sangeetha", "A2B", "A2B", "Junior Kuppanna"],
    "FoodPreference": ["veg", "Non-Veg", "Vegetarian", "Veg", "non-Veg", "Veg", "Vegetarian", "Non-Veg", "veg", "Non-Veg", "Veg"],
    "Bill": [850, 1200, 950, 1500, 1800, 2200, -500, 1100, 1350, 1350, 2500],
    "NoOfPax": [2, 4, 3, 2, 5, -1, 3, 2, 4, 4, -5],
    "EstimatedSalary": [25000, 45000, 38000, 52000, 65000, 72000, -10000, 55000, 48000, 48000, 80000]
}

df = pd.DataFrame(data)

print(df)

df.drop_duplicates(inplace=True)

df.reset_index(drop=True, inplace=True)

df.loc[df["Bill"] < 0, "Bill"] = np.nan

df.loc[df["EstimatedSalary"] < 0, "EstimatedSalary"] = np.nan

df.loc[df["NoOfPax"] < 1, "NoOfPax"] = np.nan

df.loc[
    (df["Rating(1-5)"] < 1) | (df["Rating(1-5)"] > 5),
    "Rating(1-5)"
] = np.nan

df["FoodPreference"] = df["FoodPreference"].replace(
    ["veg", "Vegetarian"], "Veg"
)

df["FoodPreference"] = df["FoodPreference"].replace(
    ["non-Veg", "non-veg"], "Non-Veg"
)

df["Bill"] = df["Bill"].fillna(df["Bill"].mean())

df["EstimatedSalary"] = df["EstimatedSalary"].fillna(
    df["EstimatedSalary"].mean()
)

df["NoOfPax"] = df["NoOfPax"].fillna(
    df["NoOfPax"].median()
)

df["Rating(1-5)"] = df["Rating(1-5)"].fillna(
    df["Rating(1-5)"].median()
)

print("\nCleaned Dataset:")
print(df)
```
