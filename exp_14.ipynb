```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Create sample website traffic data
dates = pd.date_range(start='2018-08-01', periods=400, freq='D')
np.random.seed(10)

page_views = np.random.randint(30, 80, 400).astype(float)

# Add traffic variations and spikes
for i in range(400):
    if 180 < i < 280:
        page_views[i] += np.random.randint(50, 250)
    if i % 35 == 0:
        page_views[i] += np.random.randint(500, 1000)

# Plot a single line graph
plt.figure(figsize=(12, 3))
plt.plot(dates, page_views, color='blue', linewidth=1)

plt.ylabel('Page Views')
plt.xlabel('Date')
plt.tight_layout()
plt.show()

```
