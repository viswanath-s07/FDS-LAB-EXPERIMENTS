```python
import numpy as np
import pandas as pd

# 1. Create a DataFrame using a list
data = [[1, 'Smith', 50000],
        [2, 'Jones', 60000]]

df = pd.DataFrame(data)
print("Initial DataFrame:")
print(df)

# 2. Assign column names
df.columns = ['Empd', 'Name', 'Salary']
print("\nDataFrame with Column Names:")
print(df)

# 3. Display DataFrame information
print("\nDataFrame Information:")
df.info()

# 4. Create a sample Startup dataset
startup_data = {
    'R&D Spend': [165349.20, 162597.70, 153441.51, 144372.41, 142107.34],
    'Administration': [136897.80, 151377.59, 101145.55, 118671.85, 91391.77],
    'Marketing Spend': [471784.10, 443898.53, 407934.54, 383199.62, 366168.42],
    'State': ['New York', 'California', 'Florida', 'New York', 'Florida'],
    'Profit': [192261.83, 191792.06, 191050.39, 182901.99, 166187.94]
}

startups = pd.DataFrame(startup_data)

print("\nStartup Dataset Information:")
startups.info()

print("\nFirst Five Startup Records:")
print(startups.head())

print("\nLast Five Startup Records:")
print(startups.tail())

# 5. Create Employee dataset
employee_data = {
    'emp id': [1, 2, 3, 4, 5, 6, 7],
    'name': [
        'SREE VARSSINI K S',
        'SREEMATHI B',
        'SREYA G',
        'SREYASKARI MULLAPUDI',
        'SRI AKASH U G',
        'SRI HARSHAVARDHANAN R',
        'SRI HARSHAVARDHANAN R'
    ],
    'salary': [5000, 6000, 7000, 5000, 8000, 3000, 6000]
}

df = pd.DataFrame(employee_data)

# 6. Display employee records
print("\nEmployee DataFrame:")
print(df.head())

print("\nLast Five Employee Records:")
print(df.tail())

print("\nEmployee DataFrame Information:")
df.info()

# 7. Display Salary column
print("\nSalary Column:")
print(df['salary'])

print("\nType of Salary:")
print(type(df['salary']))

# 8. Calculate statistical values
print("\nMean Salary:")
print(df['salary'].mean())

print("\nMedian Salary:")
print(df['salary'].median())

print("\nMode Salary:")
print(df['salary'].mode())

print("\nSalary Variance:")
print(df['salary'].var())

print("\nSalary Standard Deviation:")
print(df['salary'].std())

# 9. Statistical description
print("\nNumerical Description:")
print(df.describe())

print("\nComplete Description:")
print(df.describe(include='all'))

# 10. Display column names
empCol = df.columns
print("\nEmployee Column Names:")
print(empCol)

# 11. Convert DataFrame to NumPy array
emparray = df.values
print("\nEmployee NumPy Array:")
print(emparray)

# 12. Convert NumPy array back to DataFrame
employee_DF = pd.DataFrame(emparray, columns=empCol)

print("\nFinal Employee DataFrame:")
print(employee_DF)
```
