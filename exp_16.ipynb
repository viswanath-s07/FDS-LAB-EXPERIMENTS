```python
import numpy as np
import matplotlib.pyplot as plt

# Generate population
population = np.random.normal(50, 10, 100000)
population_mean = np.mean(population)

# Define sample sizes
sample_sizes = [30, 50, 100]
num_samples = 1000

sample_means = {}

# Draw random samples and calculate means
for size in sample_sizes:
    sample_means[size] = []
    for _ in range(num_samples):
        sample = np.random.choice(population, size=size, replace=False)
        sample_means[size].append(np.mean(sample))

# Plot sampling distributions
plt.figure(figsize=(10, 8))

for i, size in enumerate(sample_sizes):
    plt.subplot(3, 1, i + 1)
    plt.hist(sample_means[size], bins=30, color='skyblue',
             edgecolor='black', alpha=0.7)
    plt.axvline(population_mean, color='red',
                linestyle='dashed', linewidth=2,
                label='Population Mean')
    plt.title(f'Sample Size = {size}')
    plt.xlabel('Sample Mean')
    plt.ylabel('Frequency')
    plt.legend()

plt.tight_layout()
plt.show()
```
