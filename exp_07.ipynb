```python
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns

# Custom input data - changed values
data = {
    'total_bill': [21.5, 14.8, 36.2, 19.7, 52.3, 26.4, 33.8, 11.9, 41.6, 29.5],
    'tip': [3.5, 2.5, 6.0, 3.0, 8.5, 4.0, 5.5, 2.0, 7.0, 4.5],
    'sex': ['Female', 'Male', 'Female', 'Male', 'Female',
            'Male', 'Female', 'Male', 'Female', 'Male'],
    'day': ['Thur', 'Fri', 'Sat', 'Sun', 'Fri',
            'Sat', 'Sun', 'Thur', 'Sat', 'Sun'],
    'time': ['Lunch', 'Dinner', 'Dinner', 'Dinner', 'Dinner',
             'Lunch', 'Dinner', 'Lunch', 'Dinner', 'Dinner'],
    'size': [2, 3, 2, 4, 6, 3, 4, 2, 5, 3]
}

# Create DataFrame
tips = pd.DataFrame(data)

print("First 5 rows:")
print(tips.head())

# Joint plot
sns.jointplot(data=tips, x='tip', y='total_bill', kind='reg')
plt.show()

# Count of Lunch/Dinner
print("\nTime Value Counts:")
print(tips['time'].value_counts())

# Pair plot
sns.pairplot(tips, hue='day')
plt.show()

# Correlation heatmap
sns.heatmap(
    tips.corr(numeric_only=True),
    annot=True,
    cmap='coolwarm'
)
plt.title('Correlation Heatmap')
plt.show()

# Boxplot - Total Bill
sns.boxplot(data=tips, x='total_bill')
plt.title('Total Bill Distribution')
plt.show()

# Boxplot - Tip
sns.boxplot(data=tips, x='tip')
plt.title('Tip Distribution')
plt.show()

# Count plot - Day
sns.countplot(data=tips, x='day')
plt.title('Count by Day')
plt.show()

# Count plot - Sex
sns.countplot(data=tips, x='sex')
plt.title('Count by Sex')
plt.show()

# Pie chart - Sex
tips['sex'].value_counts().plot(
    kind='pie',
    autopct='%1.1f%%'
)
plt.title('Sex Distribution')
plt.ylabel('')
plt.show()

# Bar chart - Sex
tips['sex'].value_counts().plot(kind='bar')
plt.title('Count by Sex')
plt.xlabel('Sex')
plt.ylabel('Count')
plt.show()

# Dinner count by day
sns.countplot(
    data=tips[tips['time'] == 'Dinner'],
    x='day'
)
plt.title('Dinner Count by Day')
plt.show()
```
