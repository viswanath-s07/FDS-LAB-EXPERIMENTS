```python
import numpy as np
import matplotlib.pyplot as plt

array = np.array([10, 12, 15, 18, 20, 22, 24, 25,
                  27, 28, 30, 32, 35, 38, 75, 80])

print("Given Array:")
print(array)

print("Mean:", array.mean())

print("25th Percentile:", np.percentile(array, 25))
print("50th Percentile:", np.percentile(array, 50))
print("75th Percentile:", np.percentile(array, 75))
print("100th Percentile:", np.percentile(array, 100))

def outDetection(array):
    Q1, Q3 = np.percentile(array, [25, 75])
    IQR = Q3 - Q1

    lr = Q1 - (1.5 * IQR)
    ur = Q3 + (1.5 * IQR)

    return lr, ur

lr, ur = outDetection(array)

print("Lower Range:", lr)
print("Upper Range:", ur)

outliers = array[(array < lr) | (array > ur)]

print("Outliers:")
print(outliers)

new_array = array[(array >= lr) & (array <= ur)]

print("Array after removing outliers:")
print(new_array)

plt.hist(array)
plt.title("Before Outlier Removal")
plt.xlabel("Values")
plt.ylabel("Frequency")
plt.show()

lr1, ur1 = outDetection(new_array)

print("New Lower Range:", lr1)
print("New Upper Range:", ur1)

final_array = new_array[(new_array >= lr1) & (new_array <= ur1)]

print("Final Array:")
print(final_array)

plt.hist(final_array)
plt.title("After Outlier Removal")
plt.xlabel("Values")
plt.ylabel("Frequency")
plt.show()
```
