```python
import numpy as np
import pandas as pd

data = {
    "Country": [
        "India", "USA", "Canada", "India", "Australia",
        "Canada", "USA", "India", np.nan, "Australia"
    ],

    "Age": [
        25, 32, 29, 41, 36, 28, np.nan, 45, 39, 34
    ],

    "Salary": [
        35000, 62000, 58000, 72000, np.nan,
        54000, 68000, 81000, 59000, 65000
    ],

    "Purchased": [
        "Yes", "No", "Yes", "Yes", "No",
        "No", "Yes", "No", "Yes", "No"
    ]
}

df = pd.DataFrame(data)

print("Original Dataset:")

display(df)

print("Dataset Information:")

df.info()

print("Mode of Country:")

print(df["Country"].mode())

df["Country"] = df["Country"].fillna(
    df["Country"].mode()[0]
)

df["Age"] = df["Age"].fillna(
    df["Age"].median()
)

df["Salary"] = df["Salary"].fillna(
    round(df["Salary"].mean())
)

print("After Data Preprocessing:")

display(df)
```
