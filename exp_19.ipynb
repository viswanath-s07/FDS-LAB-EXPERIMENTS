```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.datasets import load_iris

# Load Iris dataset without a CSV file
flower_data = load_iris()

iris_df = pd.DataFrame(
    flower_data.data,
    columns=flower_data.feature_names
)
iris_df['variety'] = [
    flower_data.target_names[i]
    for i in flower_data.target
]

# Display dataset information
print("Dataset Information:")
iris_df.info()

print("\nStatistical Summary:")
print(iris_df.describe())

print("\nFlower Variety Counts:")
print(iris_df['variety'].value_counts())

# Count plot
plt.figure(figsize=(7, 5))
sns.countplot(data=iris_df, x='variety')
plt.title("Flower Count by Variety")
plt.show()

# Scatter plot 1
plt.figure(figsize=(7, 5))
sns.scatterplot(
    data=iris_df,
    x='sepal length (cm)',
    y='sepal width (cm)',
    hue='variety'
)
plt.title("Sepal Length vs Sepal Width")
plt.show()

# Scatter plot 2
plt.figure(figsize=(7, 5))
sns.scatterplot(
    data=iris_df,
    x='petal length (cm)',
    y='petal width (cm)',
    hue='variety'
)
plt.title("Petal Length vs Petal Width")
plt.show()

# Pair plot
sns.pairplot(iris_df, hue='variety', diag_kind='hist')
plt.show()

# Histograms
for feature in flower_data.feature_names:
    plt.figure(figsize=(7, 4))
    sns.histplot(data=iris_df, x=feature, kde=True)
    plt.title(f"Distribution of {feature}")
    plt.show()
```
