```python
import pandas as pd
import matplotlib.pyplot as plt

data = {
    'Year': list(range(2015, 2026)),
    'Marks': [55, 58, 62, 65, 68, 72, 75, 78, 82, 85, 88]
}

df = pd.DataFrame(data)

plt.plot(df['Year'], df['Marks'], marker='o')
plt.title('Student Marks Improvement')
plt.xlabel('Year')
plt.ylabel('Marks')
plt.show()
```

```python
import matplotlib.pyplot as plt

subjects = ['Maths', 'Science', 'English', 'History', 'Computer']
students = [85, 70, 95, 60, 110]

plt.bar(subjects, students)
plt.title('Students by Subject')
plt.xlabel('Subject')
plt.ylabel('Number of Students')
plt.show()
```

```python
import pandas as pd

# Structured data
structured_data = pd.DataFrame({
    'Employee_ID': [201, 202, 203],
    'Employee': ['Priya', 'Vijay', 'Sneha'],
    'Salary': [45000, 52000, 48000]
})

print("Structured Data:\n", structured_data)

# Unstructured data
unstructured_data = "Priya is an excellent employee who has strong communication and teamwork skills."

print("\nUnstructured Data:\n", unstructured_data)

# Semi-structured data
semi_structured_data = {
    'Employee_ID': 201,
    'Employee': 'Priya',
    'Details': {
        'Department': 'Human Resources',
        'Experience': '5 Years'
    }
}

print("\nSemi-structured Data:\n", semi_structured_data)
```

```python

from cryptography.fernet import Fernet

key = Fernet.generate_key()
cipher_suite = Fernet(key)

plain_text = b"Artificial Intelligence and Machine Learning"
cipher_text = cipher_suite.encrypt(plain_text)

# Decrypt data
decrypted_text = cipher_suite.decrypt(cipher_text)

print("Original Data:", plain_text)
print("Encrypted Data:", cipher_text)
print("Decrypted Data:", decrypted_text)
```
